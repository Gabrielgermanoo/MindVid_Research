```python
import os


def download_video(save_directory, link, urls, csv_path, key, cont):
    ydl_opts = {
        "format": "bestaudio/best",
        "postprocessors": [
            {
                "key": "FFmpegExtractAudio",
                "preferredcodec": "wav",
                "preferredquality": "192",
            }
        ],
        "outtmpl": os.path.join(save_directory, f"{cont}_%(id)s.%(ext)s"),
        "headers": {
            "User-Agent": "Chrome/80.0.3987.122 Safari/537.36",
            "Referer": "https://www.tiktok.com/",
        },
        "cookiesfrombrowser": ("chrome",),
        "nocheckcertificate": True,
        "no_check_certificate": True,
        "retries": 3,
        "fragment_retries": 3,
        "extractor_retries": 3,
    }
    try:
        print(f"Downloading from: {link}")
        with youtube_dl.YoutubeDL(ydl_opts) as ydl:
            ydl.download([link])
        print(f"Download successful: {cont}")
        return True

    except youtube_dl.utils.DownloadError as e:
        print(f"Download error for {link}: {e}")
        # Remove o link do DataFrame se falhar
        try:
            urls = urls[urls["Link"] != link]
            urls = urls.reset_index(drop=True)
            urls.to_csv(os.path.join(csv_path, f"{key}.csv"), index=False)
            print("Removed failed link from CSV")
        except Exception as csv_error:
            print(f"Could not update CSV: {csv_error}")
        return False

    except Exception as e:
        print(f"Unexpected error for {link}: {e}")
        return False
```

```python
key = "anorexia_homens"
link = "https://www.instagram.com/reel/Czw6poXrWEv/?igsh=cmQ3Ymg1dWY4aDl2"
save_directory = f"downloads/{key}"
csv_path = f'./CSV/{key}'
df = pd.read_csv(f'{csv_path}/{key}.csv')
cont = 9
download_video(save_directory, link, df, csv_path, key, cont)
```

```python
for key, hashtags in hashtags_list.items():
    csv_path = f'../CSV/{key}'
    save_directory = f'../Videos/{key}'
    cont = 0

    # Carregar o arquivo CSV correspondente ao tema
    df = pd.read_csv(f'{csv_path}/{key}.csv')

    # Iterar sobre os links e baixar os vídeos
    for link in df['Link']:
        download_video(save_directory, link, df, csv_path, key, cont)
        print(f'Downloaded {key} video {cont + 1}')
        cont += 1
```
