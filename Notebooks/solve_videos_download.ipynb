```python
import yt_dlp as youtube_dl
import pandas as pd
```

```python
hashtags_list = {
    # "ansiedade": ["#ansiedade"],
    # "depressao": ["#depressao"],
    # "TDAH": ["#TDAH"],
    # "TEA": ["#autismo"],
    # "TOC": ["#TOC"],
    # "TEPT": ["#TEPT"],
    # "TBP": ["#TBP"],
    # "borderline": ["#borderline"],
    "anorexia_homem": ["#anorexiahomem"]
}
```

```python
# download method
def download_video(save_directory, link, urls, csv_path, key, cont):
    ydl_opts = {
        "format": "bestaudio/best",
        "postprocessors": [
            {
                "key": "FFmpegExtractAudio",
                "preferredcodec": "wav",
                "preferredquality": "192",
            }
        ],
        "outtmpl": save_directory + "/" + str(cont) + "_%(id)s.%(ext)s",
        "headers": {
            "User-Agent": "Instagram 219.0.0.12.117 Android",
        },
        "cookiesfrombrowser": ("chrome",),
    }
    try:
        with youtube_dl.YoutubeDL(ydl_opts) as ydl:
            ydl.download([link])
        cont += 1
    except youtube_dl.utils.DownloadError:
        urls = urls.drop(cont)
        urls = urls.reset_index(drop=True)
        urls.to_csv(csv_path + "/" + key + ".csv", index=False, header=False)
```

```python
key = "anorexia_homens"
link = "https://www.instagram.com/reel/Czw6poXrWEv/?igsh=cmQ3Ymg1dWY4aDl2"
save_directory = f"downloads/{key}"
csv_path = f'./CSV/{key}'
df = pd.read_csv(f'{csv_path}/{key}.csv')
cont = 9
download_video(save_directory, link, df, csv_path, key, cont)
```

```python
for key, hashtags in hashtags_list.items():
    csv_path = f'../CSV/{key}'
    save_directory = f'../Videos/{key}'
    cont = 0

    # Carregar o arquivo CSV correspondente ao tema
    df = pd.read_csv(f'{csv_path}/{key}.csv')

    # Iterar sobre os links e baixar os vídeos
    for link in df['Link']:
        download_video(save_directory, link, df, csv_path, key, cont)
        print(f'Downloaded {key} video {cont + 1}')
        cont += 1
```
